```python
## SINGLE TEST FOR MULTIPLE SENTECES
import nltk
import pandas as pd
nltk.download('punkt')
from nltk.tokenize import sent_tokenize

def conformance_test(text):
    """
    Check if the text consists of multiple phrases.

    Parameters:
    text (str): The text to be checked.

    Returns:
    bool: True if the text consists of multiple phrases, False otherwise.
    """
    # Tokenize the text into sentences
    sentences = sent_tokenize(text)
    
    # Count the number of sentences
    return len(sentences) > 1  # Return True if more than one sentence, False otherwise

# Example usage
text = "This is a test sentence. This is another sentence."
result = conformance_test(text)
print(f"Does the text consist of multiple phrases? {result}")

# # Test with single phrase
# single_phrase_text = "This is a single sentence."
# single_phrase_result = conformance_test(single_phrase_text)
# print(f"Does the single phrase text consist of multiple phrases? {single_phrase_result}")
```

```python
##UNIT TEST FOR CONFORMANCE TESTING

import os
import pandas as pd
from nltk.tokenize import sent_tokenize

# Define the conformance test function
def conformance_test(text):
    """
    Check if the text consists of multiple phrases.

    Parameters:
    text (str): The text to be checked.

    Returns:
    bool: True if the text consists of multiple phrases, False otherwise.
    """
    # Tokenize the text into sentences
    sentences = sent_tokenize(text)
    
    # Count the number of sentences
    return len(sentences) > 1  # Return True if more than one sentence, False otherwise

# Define the folder path and filename
folder_path = 'Desktop/Notebook/Shamus/ipex+tcm+AMX+bf16/'
file_name = 'inference_n=14.csv'

# Concatenate to get the full file path
file_path = os.path.join(folder_path, file_name)

# Read the CSV file
df = pd.read_csv(file_path)

# Ensure 'Generated Output' is filled with empty strings where there are NaNs
df['Generated Output'] = df['Generated Output'].fillna('')

# Apply the combined conformance test
def combined_conformance_test(text):
    """
    Check if the text passes both the conformance test and does not contain ':'.

    Parameters:
    text (str): The text to be checked.

    Returns:
    bool: True if the text passes both tests, False otherwise.
    """
    return conformance_test(text) and (':' not in text)

df['Conformance'] = df['Generated Output'].apply(combined_conformance_test)

# Calculate the final conformance test value
num_true = df['Conformance'].sum()
total_rows = len(df)
conformance_test_value = num_true / total_rows

print(f"Final conformance test value: {conformance_test_value}")

# Save the rows containing ":" to one CSV file
colon_rows = df[~df['Conformance']]
print(f"{len(colon_rows)} rows not passing conformance test have been saved in colon_rows.csv.")

# Save the rows passing the conformance test to another CSV file
no_colon_rows = df[df['Conformance']]
print(f"{len(no_colon_rows)} rows passing conformance test have been saved in no_colon_rows.csv.")
```
